```python
import numpy as np
import matplotlib.pyplot as plt

def objective_function(x):
    return x * np.sin(x) + 0.1 * x**2 + 10 * np.cos(2*x)


n_wolves = 10 
max_iter = 16
lb, ub = -30, 30
positions = np.random.uniform(lb, ub, n_wolves)

x_plot = np.linspace(lb, ub, 1000)
y_plot = objective_function(x_plot)

for t in range(max_iter):
    fitness = objective_function(positions)
    sorted_idx = np.argsort(fitness)
    
   
    alpha_pos = positions[sorted_idx[0]]
    beta_pos = positions[sorted_idx[1]]
    delta_pos = positions[sorted_idx[2]]
    
    
    a = 2 - t * (2 / max_iter) 
    
    
    print(f"\n{'='*60}")
    print(f" ITERATION {t+1}")
    print(f"{'='*60}")
    print(f"Alpha (Rank 1): x = {alpha_pos:.4f}, f(x) = {fitness[sorted_idx[0]]:.4f}")
    print(f"Beta  (Rank 2): x = {beta_pos:.4f}")
    print(f"Delta (Rank 3): x = {delta_pos:.4f}")
    print(f"Current 'a' parameter: {a:.4f}")
    
    
    sample_wolf_idx = sorted_idx[-1] 
    curr_p_sample = positions[sample_wolf_idx]
    
    new_positions = []
    plt.figure(figsize=(15, 6))
    plt.plot(x_plot, y_plot, 'b-', alpha=0.3)

    for i in range(n_wolves):
        curr_p = positions[i]
        
        
        def get_X_update(target, name, show_details=False):
            r1, r2 = np.random.rand(), np.random.rand()
            A = 2 * a * r1 - a
            C = 2 * r2
            D = abs(C * target - curr_p)
            X = target - A * D
            if show_details:
                print(f"\nCalculation for {name}:")
                print(f"  r1={r1:.4f}, r2={r2:.4f} -> A={A:.4f}, C={C:.4f}")
                print(f"  D = |({C:.4f} * {target:.4f}) - ({curr_p:.4f})| = {D:.4f}")
                print(f"  X_suggested = {target:.4f} - ({A:.4f} * {D:.4f}) = {X:.4f}")
            return X

       
        is_sample = (i == sample_wolf_idx)
        if is_sample:
            print(f"\nDETAILED TRACE FOR WOLF #{i} (Current X: {curr_p:.4f}):")
        
        X1 = get_X_update(alpha_pos, "Alpha", is_sample)
        X2 = get_X_update(beta_pos, "Beta", is_sample)
        X3 = get_X_update(delta_pos, "Delta", is_sample)
        
        next_pos = (X1 + X2 + X3) / 3
        new_positions.append(next_pos)
        
        if is_sample:
            print(f"\nFinal Averaged Position for Wolf #{i}:")
            print(f"  X(t+1) = ({X1:.4f} + {X2:.4f} + {X3:.4f}) / 3 = {next_pos:.4f}")

        
        plt.annotate('', xy=(next_pos, objective_function(next_pos)), 
                     xytext=(curr_p, objective_function(curr_p)),
                     arrowprops=dict(arrowstyle="->", color='gray', alpha=0.4))

    
    plt.scatter(positions[sorted_idx[3:]], objective_function(positions[sorted_idx[3:]]), c='black', label='Omegas', alpha=0.6)
    plt.scatter(alpha_pos, objective_function(alpha_pos), c='gold', marker='*', s=400, label='Alpha', zorder=5)
    plt.scatter(beta_pos, objective_function(beta_pos), c='silver', marker='P', s=200, label='Beta', zorder=4)
    plt.scatter(delta_pos, objective_function(delta_pos), c='brown', marker='D', s=150, label='Delta', zorder=3)

    plt.title(f"GWO Iteration {t+1}")
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()

    positions = np.clip(new_positions, lb, ub)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_vw_weights(max_iter):
    iterations = np.arange(1, max_iter + 1)
    w1_list, w2_list, w3_list = [], [], []
    const_factor = (2 / np.pi) * np.arccos(1/3)

    print(f"{'Iter':<10} | {'w1 (Alpha)':<12} | {'w2 (Beta)':<12} | {'w3 (Delta)':<12}")
    print("-" * 55)

    for it in iterations:
        phi = 0.5 * np.arctan(it)
        theta = const_factor * np.arctan(it)

        
        w1 = np.cos(theta)
        w2 = 0.5 * np.sin(theta) * np.cos(phi)
        w3 = 1 - w1 - w2

        w1_list.append(w1)
        w2_list.append(w2)
        w3_list.append(w3)

        print(f"{it:<10} | {w1:<12.4f} | {w2:<12.4f} | {w3:<12.4f}")

    return iterations, w1_list, w2_list, w3_list


iters, w1, w2, w3 = calculate_vw_weights(20)


plt.figure(figsize=(10, 6))
plt.plot(iters, w1, 'gold', label='w1 (Alpha Influence)', linewidth=2, marker='*')
plt.plot(iters, w2, 'silver', label='w2 (Beta Influence)', linewidth=2, marker='P')
plt.plot(iters, w3, 'brown', label='w3 (Delta Influence)', linewidth=2, marker='D')

plt.title('Dynamic Weight Evolution in VW-GWO (Trigonometric Approach)')
plt.xlabel('Iteration (it)')
plt.ylabel('Weight Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(iters)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Objective Function (Rastrigin) ---
def objective_function(x):
    return np.sum(x**2 - 10 * np.cos(2 * np.pi * x) + 10)


def run_standard_gwo(positions, max_iter, lb, ub, dim, n_wolves):
    pos = positions.copy()
    score = np.array([objective_function(ind) for ind in pos])
    
   
    alpha_pos, beta_pos, delta_pos = np.zeros(dim), np.zeros(dim), np.zeros(dim)
    alpha_score, beta_score, delta_score = float("inf"), float("inf"), float("inf")
    
    convergence_curve = []

    for t in range(max_iter):
        
        sorted_idx = np.argsort(score)
        alpha_pos, alpha_score = pos[sorted_idx[0]].copy(), score[sorted_idx[0]]
        beta_pos, beta_score = pos[sorted_idx[1]].copy(), score[sorted_idx[1]]
        delta_pos, delta_score = pos[sorted_idx[2]].copy(), score[sorted_idx[2]]
        
        convergence_curve.append(alpha_score)
        
        a = 2 - t * (2 / max_iter) 
        
        for i in range(n_wolves):
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            A1, C1 = 2*a*r1 - a, 2*r2
            D_alpha = np.abs(C1 * alpha_pos - pos[i])
            X1 = alpha_pos - A1 * D_alpha
            
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            A2, C2 = 2*a*r1 - a, 2*r2
            D_beta = np.abs(C2 * beta_pos - pos[i])
            X2 = beta_pos - A2 * D_beta
            
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            A3, C3 = 2*a*r1 - a, 2*r2
            D_delta = np.abs(C3 * delta_pos - pos[i])
            X3 = delta_pos - A3 * D_delta
            
            pos[i] = (X1 + X2 + X3) / 3 
            
        
        pos = np.clip(pos, lb, ub)
        score = np.array([objective_function(ind) for ind in pos])
        
    return convergence_curve

def run_vw_gwo(positions, max_iter, lb, ub, dim, n_wolves):
    pos = positions.copy()
    score = np.array([objective_function(ind) for ind in pos])
    
    alpha_pos, beta_pos, delta_pos = np.zeros(dim), np.zeros(dim), np.zeros(dim)
    alpha_score, beta_score, delta_score = float("inf"), float("inf"), float("inf")
    
    convergence_curve = []
    
    
    const_factor = (2 / np.pi) * np.arccos(1/3)

    for t in range(1, max_iter + 1): 
        sorted_idx = np.argsort(score)
        alpha_pos, alpha_score = pos[sorted_idx[0]].copy(), score[sorted_idx[0]]
        beta_pos, beta_score = pos[sorted_idx[1]].copy(), score[sorted_idx[1]]
        delta_pos, delta_score = pos[sorted_idx[2]].copy(), score[sorted_idx[2]]
        
        convergence_curve.append(alpha_score)
        
        #(VW-GWO Logic)
        phi = 0.5 * np.arctan(t)
        theta = const_factor * np.arctan(t)
        w1 = np.cos(theta)
        w2 = 0.5 * np.sin(theta) * np.cos(phi)
        w3 = 1 - w1 - w2
        
        
        a = 2 * np.exp(-t / max_iter) 
        
        for i in range(n_wolves):
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            X1 = alpha_pos - (2*a*r1 - a) * np.abs(2*r2 * alpha_pos - pos[i])
            
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            X2 = beta_pos - (2*a*r1 - a) * np.abs(2*r2 * beta_pos - pos[i])
            
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            X3 = delta_pos - (2*a*r1 - a) * np.abs(2*r2 * delta_pos - pos[i])
            
            pos[i] = w1 * X1 + w2 * X2 + w3 * X3
            
        pos = np.clip(pos, lb, ub)
        score = np.array([objective_function(ind) for ind in pos])
        
    return convergence_curve

def run_iag_gwo(positions, max_iter, lb, ub, dim, n_wolves):
    pos = positions.copy()
    score = np.array([objective_function(ind) for ind in pos])
    
    alpha_pos = np.zeros(dim)
    alpha_score = float("inf")
    
    convergence_curve = []

    for t in range(max_iter):
       
        sorted_idx = np.argsort(score)
        current_alpha = pos[sorted_idx[0]].copy()
        current_score = score[sorted_idx[0]]
        

        obl_alpha = (lb + ub) - current_alpha
        obl_score = objective_function(obl_alpha)
        
        # 2. Gaussian Mutation
        r_gauss = np.random.normal(0, 1, dim)
        mut_alpha = current_alpha * (1 + 0.1 * r_gauss) 
        mut_alpha = np.clip(mut_alpha, lb, ub)
        mut_score = objective_function(mut_alpha)
        best_candidate_pos = current_alpha
        best_candidate_score = current_score
        
        if obl_score < best_candidate_score:
            best_candidate_pos = obl_alpha
            best_candidate_score = obl_score
        if mut_score < best_candidate_score:
            best_candidate_pos = mut_alpha
            best_candidate_score = mut_score
            
      
        pos[sorted_idx[0]] = best_candidate_pos
        score[sorted_idx[0]] = best_candidate_score
        
        
        alpha_pos = best_candidate_pos
        alpha_score = best_candidate_score

        beta_pos = pos[sorted_idx[1]]
        delta_pos = pos[sorted_idx[2]]

        convergence_curve.append(alpha_score)
        
        a = 2 - t * (2 / max_iter)
        
        for i in range(n_wolves):
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            X1 = alpha_pos - (2*a*r1 - a) * np.abs(2*r2 * alpha_pos - pos[i])
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            X2 = beta_pos - (2*a*r1 - a) * np.abs(2*r2 * beta_pos - pos[i])
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            X3 = delta_pos - (2*a*r1 - a) * np.abs(2*r2 * delta_pos - pos[i])
            
            pos[i] = (X1 + X2 + X3) / 3
            
        pos = np.clip(pos, lb, ub)
        score = np.array([objective_function(ind) for ind in pos])
        
    return convergence_curve

# --- 3. Main Execution ---
n_wolves = 30
dim = 10
max_iter = 100
lb, ub = -5.12, 5.12 


np.random.seed(42) 
initial_population = np.random.uniform(lb, ub, (n_wolves, dim))

print("Running Standard GWO...")
curve_gwo = run_standard_gwo(initial_population, max_iter, lb, ub, dim, n_wolves)

print("Running VW-GWO (Variable Weight)...")
curve_vw = run_vw_gwo(initial_population, max_iter, lb, ub, dim, n_wolves)

print("Running IAgGWO (Improved Alpha)...")
curve_iag = run_iag_gwo(initial_population, max_iter, lb, ub, dim, n_wolves)

# --- 4. Plotting ---
plt.figure(figsize=(10, 6))
iterations = range(1, max_iter + 1)

plt.plot(iterations, curve_gwo, 'k--', label='Standard GWO', linewidth=1.5)
plt.plot(iterations, curve_vw, 'b-', label='VW-GWO (Variable Weights)', linewidth=2)
plt.plot(iterations, curve_iag, 'r-', label='IAgGWO (Improved Alpha)', linewidth=2)

plt.yscale('log') 
plt.title('Convergence Comparison: Standard GWO vs Improved Variants\n(Rastrigin Function - 10 Dim)')
plt.xlabel('Iteration')
plt.ylabel('Best Fitness (Log Scale)')
plt.legend()
plt.grid(True, which="both", linestyle='--', alpha=0.6)
plt.show()
```
